# **Feature Engineering**

## **Creating New Features in the Dataset**

The following code snippet demonstrates the creation of new features based on existing ones in the dataset:

### 1. **Total Activity Hours**:
   - **`df['Total_Activity_Hours'] = df['Extracurricular_Hours_Per_Day'] + df['Physical_Activity_Hours_Per_Day']`**:
     - This feature represents the total amount of activity (both extracurricular and physical) a student engages in per day.
  
### 2. **Free Time Hours**:
   - **`df['Free_Time_Hours'] = 24 - (df['Study_Hours_Per_Day'] + df['Sleep_Hours_Per_Day'] + df['Total_Activity_Hours'] + df['Social_Hours_Per_Day'])`**:
     - This feature calculates the available free time a student has per day by subtracting the time spent on various activities (study, sleep, activities, socializing) from 24 hours.
  
### 3. **Categorizing GPA**:
   - **`df['GPA_Category'] = pd.cut(df['GPA'], bins=bins, labels=labels)`**:
     - The `GPA` is categorized into three bins: Low (GPA ≤ 2.0), Medium (2.0 < GPA ≤ 3.5), and High (GPA > 3.5).
     - The bins and corresponding labels are defined as:
       - Bins: `[0, 2.0, 3.5, 4.0]`
       - Labels: `['Low', 'Medium', 'High']`
  
### 4. **Interaction Term: Study Time vs GPA**:
   - **`df['Study_GPA_Ratio'] = df['Study_Hours_Per_Day'] / (df['GPA'] + 1e-5)`**:
     - This interaction term represents the ratio of study hours to GPA, providing insight into the relationship between study time and academic performance.
     - A small constant (`1e-5`) is added to avoid division by zero.

### 5. **Interaction Term: Sleep Time vs GPA**:
   - **`df['Sleep_GPA_Ratio'] = df['Sleep_Hours_Per_Day'] / (df['GPA'] + 1e-5)`**:
     - Similar to the study-to-GPA ratio, this feature captures the relationship between sleep hours and GPA.

### 6. **Binary Feature for Stress**:
   - **`df['Is_Stressed'] = df['Stress_Level'].map({'Low': 0, 'Moderate': 1, 'High': 1})`**:
     - Converts the `Stress_Level` feature into a binary format, where `Low` stress is mapped to `0`, and `Moderate` and `High` stress levels are mapped to `1`.

### 7. **Removing Duplicate Rows**:
   - **`df = df.drop_duplicates()`**:
     - Removes any duplicate rows in the dataset, ensuring the data is unique and clean.

```python
# Create new features
df['Total_Activity_Hours'] = df['Extracurricular_Hours_Per_Day'] + df['Physical_Activity_Hours_Per_Day']
df['Free_Time_Hours'] = 24 - (df['Study_Hours_Per_Day'] + df['Sleep_Hours_Per_Day'] + df['Total_Activity_Hours'] + df['Social_Hours_Per_Day'])

# Categorize GPA into bins (e.g., Low, Medium, High)
bins = [0, 2.0, 3.5, 4.0]  # Define GPA ranges
labels = ['Low', 'Medium', 'High']
df['GPA_Category'] = pd.cut(df['GPA'], bins=bins, labels=labels)

# Interaction term: Study time vs GPA
df['Study_GPA_Ratio'] = df['Study_Hours_Per_Day'] / (df['GPA'] + 1e-5)  # Avoid division by zero

# Interaction term: Sleep time vs GPA
df['Sleep_GPA_Ratio'] = df['Sleep_Hours_Per_Day'] / (df['GPA'] + 1e-5)  # Avoid division by zero

# Example of binary feature
df['Is_Stressed'] = df['Stress_Level'].map({'Low': 0, 'Moderate': 1, 'High': 1})

df = df.drop_duplicates()
```

## **Identifying and Handling Outliers**

The following code snippet demonstrates how to identify outliers using z-scores and handle them by capping extreme values using the Interquartile Range (IQR) method:

### 1. **Identifying Outliers Using Z-Scores**:
   - **`from scipy.stats import zscore`**:
     - Imports the `zscore` function from SciPy, which calculates the z-scores for numeric features.
   - **`numeric_features = df.select_dtypes(include=['float64', 'int64']).columns`**:
     - Selects all numeric columns from the DataFrame to analyze for outliers.
   - **`z_scores = np.abs(zscore(df[numeric_features]))`**:
     - Computes the absolute z-scores for all numeric features. Z-scores indicate how many standard deviations a data point is from the mean.
   - **`outliers = (z_scores > 3).any(axis=1)`**:
     - Flags rows with z-scores greater than 3 as outliers. A z-score above 3 (or below -3) is commonly considered an outlier.

### 2. **Capping Outliers Using the IQR Method**:
   - **`for col in numeric_features:`**:
     - Loops through each numeric column to handle outliers.
   - **`q1, q3 = df[col].quantile([0.25, 0.75])`**:
     - Calculates the first (Q1) and third (Q3) quartiles for each numeric feature.
   - **`iqr = q3 - q1`**:
     - Computes the Interquartile Range (IQR), which represents the spread of the middle 50% of the data.
   - **`lower_bound = q1 - 1.5 * iqr`** and **`upper_bound = q3 + 1.5 * iqr`**:
     - Defines the lower and upper bounds for acceptable values. Any value outside this range is considered an outlier.
   - **`df[col] = np.clip(df[col], lower_bound, upper_bound)`**:
     - Caps the outlier values to the lower and upper bounds, effectively reducing extreme values to the acceptable range.

```python
# Identify outliers using z-scores
from scipy.stats import zscore
numeric_features = df.select_dtypes(include=['float64', 'int64']).columns

# Calculate z-scores
z_scores = np.abs(zscore(df[numeric_features]))
outliers = (z_scores > 3).any(axis=1)  # Flag rows with z-scores > 3

# Cap outliers (for each column)
for col in numeric_features:
    q1, q3 = df[col].quantile([0.25, 0.75])
    iqr = q3 - q1
    lower_bound = q1 - 1.5 * iqr
    upper_bound = q3 + 1.5 * iqr
    df[col] = np.clip(df[col], lower_bound, upper_bound)
```

## **Model Training and Evaluation for Different Test-Train Split Ratios**

The following code trains and evaluates two models (Linear Regression and Random Forest Regressor) using different train-test split ratios:

### 1. **Preparing Data**:
   - **`X = df.drop(columns=['GPA'])`**:
     - Drops the target variable `GPA` from the feature set `X`.
   - **`y = df['GPA']`**:
     - Defines `y` as the target variable, which is `GPA`.

### 2. **Encoding `Stress_Level`**:
   - **`X['Stress_Level'] = X['Stress_Level'].map({'Low': 0, 'Moderate': 1, 'High': 2})`**:
     - Encodes the `Stress_Level` feature into numerical values (0, 1, 2) to make it suitable for machine learning models.

### 3. **Standardizing Numeric Features**:
   - **`scaler = StandardScaler()`**:
     - Initializes the `StandardScaler` to standardize numeric features.
   - **`X_scaled = scaler.fit_transform(X)`**:
     - Standardizes the features in `X`, transforming them to have zero mean and unit variance.

### 4. **Defining Split Ratios**:
   - **`split_ratios = [0.2, 0.3, 0.4]`**:
     - Defines the test size ratios for train-test splits: 80-20, 70-30, and 60-40 splits.

### 5. **Training and Evaluation**:
   - For each split ratio, the following steps are performed:
     - **`train_test_split(X_scaled, y, test_size=ratio, random_state=42)`**:
       - Splits the dataset into training and testing sets based on the current ratio.
     - **Linear Regression Model**:
       - **`linear_model = LinearRegression()`**: Initializes the Linear Regression model.
       - **`linear_model.fit(X_train, y_train)`**: Trains the model on the training data.
       - **`y_pred_lr = linear_model.predict(X_test)`**: Makes predictions on the test data.
       - **`mean_squared_error()` and `r2_score()`**: Evaluates the performance of the model using Mean Squared Error (MSE) and R² score.
     - **Random Forest Regressor Model**:
       - **`rf_model = RandomForestRegressor(random_state=42, n_estimators=100)`**: Initializes the Random Forest Regressor model with 100 trees.
       - **`rf_model.fit(X_train, y_train)`**: Trains the model on the training data.
       - **`y_pred_rf = rf_model.predict(X_test)`**: Makes predictions on the test data.
       - **`mean_squared_error()` and `r2_score()`**: Evaluates the performance of the model using MSE and R² score.
   - The results (MSE and R² scores) are stored for both models and each split ratio.

### 6. **Output**:
   - The performance of both models (Linear Regression and Random Forest) is printed for each test size ratio:
     - Mean Squared Error (MSE)
     - R² Score

```python
# Define different test-train split sizes
split_ratios = [0.2, 0.3, 0.4]  # 80-20, 70-30, 60-40 splits

# Prepare data
X = df.drop(columns=['GPA'])
y = df['GPA']

# Encode Stress_Level
X['Stress_Level'] = X['Stress_Level'].map({'Low': 0, 'Moderate': 1, 'High': 2})

# Standardize numeric features
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

# Initialize results dictionary
results = {}

# Train models for each split ratio
for ratio in split_ratios:
    print(f"\nTraining with test size = {ratio}")

    # Split data
    X_train, X_test, y_train, y_test = train_test_split(X_scaled, y, test_size=ratio, random_state=42)

    # Linear Regression
    linear_model = LinearRegression()
    linear_model.fit(X_train, y_train)
    y_pred_lr = linear_model.predict(X_test)

    mse_lr = mean_squared_error(y_test, y_pred_lr)
    r2_lr = r2_score(y_test, y_pred_lr)

    # Random Forest Regressor
    rf_model = RandomForestRegressor(random_state=42, n_estimators=100)
    rf_model.fit(X_train, y_train)
    y_pred_rf = rf_model.predict(X_test)

    mse_rf = mean_squared_error(y_test, y_pred_rf)
    r2_rf = r2_score(y_test, y_pred_rf)

    # Store results
    results[ratio] = {
        'Linear Regression': {'MSE': mse_lr, 'R2': r2_lr},
        'Random Forest': {'MSE': mse_rf, 'R2': r2_rf}
    }

    # Print results
    print("\nLinear Regression Performance:")
    print(f"Mean Squared Error: {mse_lr:.4f}, R² Score: {r2_lr:.4f}")

    print("\nRandom Forest Performance:")
    print(f"Mean Squared Error: {mse_rf:.4f}, R² Score: {r2_rf:.4f}")
```

## **Linear Regression and Feature Significance Testing**

This code performs feature selection and significance testing using the linear regression model and the F-test (`f_regression`) to evaluate the importance of each predictor variable.

### **Null and Alternative Hypotheses**:
- **Null Hypothesis (H₀)**: There is no significant relationship between the predictor variable and the target variable (i.e., the coefficient of the predictor is zero).
- **Alternative Hypothesis (H₁)**: There is a significant relationship between the predictor variable and the target variable (i.e., the coefficient of the predictor is not zero).

### 1. **Fit a Linear Regression Model**:
   - **`model = LinearRegression()`**:
     - Initializes the Linear Regression model.
   - **`model.fit(X_scaled, y)`**:
     - Trains the model using the scaled features (`X_scaled`) and the target variable (`y`).

### 2. **Get Coefficients and Standard Error**:
   - **`coefficients = model.coef_`**:

## **ANOVA Test for Model Significance**

This code performs an **ANOVA (Analysis of Variance)** test to evaluate the significance of the regression model in predicting the target variable (`GPA`). It compares the full model (with all features) against a baseline model (mean-only model).

### 1. **Calculate Residuals for Full Model**:
   - **`y_pred = model.predict(X_scaled)`**:
     - Predicts the target variable (`y`) using the full regression model with scaled features (`X_scaled`).
   - **`residuals_full = y - y_pred`**:
     - Calculates the residuals for the full model, which represent the difference between the observed (`y`) and predicted (`y_pred`) values.

### 2. **Baseline Model (Mean-Only Model)**:
   - **`y_baseline = np.mean(y)`**:
     - Calculates the mean of the target variable (`y`), representing the baseline model where predictions are made by simply using the average value of `y`.
   - **`residuals_baseline = y - y_baseline`**:
     - Calculates the residuals for the baseline model, which is the difference between the observed values and the baseline mean.

### 3. **Calculate Sum of Squares**:
   - **`SS_full = np.sum(residuals_full^2)`**:
     - Computes the Sum of Squares for the full model (`SS_full`), which is the sum of squared residuals for the full regression model.
   - **`SS_baseline = np.sum(residuals_baseline^2)`**:
     - Computes the Sum of Squares for the baseline model (`SS_baseline`), which is the sum of squared residuals for the mean-only model.

### 4. **Degrees of Freedom**:
   - **`df_full = len(y) - len(X.columns) - 1`**:
     - Calculates the degrees of freedom for the full model, which is the total number of observations (`len(y)`) minus the number of features (`len(X.columns)`) and the intercept.
   - **`df_baseline = len(y) - 1`**:
     - Calculates the degrees of freedom for the baseline model, which is the total number of observations minus 1.

### 5. **Calculate F-statistic**:
   - **`F_stat = ((SS_baseline - SS_full) / (len(X.columns))) / (SS_full / df_full)`**:
     - Computes the F-statistic, which compares the improvement in model fit (from baseline to full model) relative to the unexplained variance.
     - The F-statistic is used to test whether the full model is significantly better than the baseline model.

### 6. **Calculate p-value**:
   - **`p_value_anova = 1 - t.cdf(F_stat, df_full)`**:
     - Computes the p-value for the F-statistic using the cumulative distribution function (CDF) of the t-distribution.
     - The p-value helps to determine whether the full model significantly explains the variance in the target variable.

### 7. **Display Results**:
   - **Print the F-statistic and p-value**:
     - Displays the calculated F-statistic and p-value.
   - **Hypothesis Testing**:
     - If the p-value is less than 0.05, the null hypothesis is rejected, indicating that the regression model significantly explains the variance in the target variable (`GPA`).
     - If the p-value is greater than 0.05, we fail to reject the null hypothesis, suggesting that the model does not explain the variance better than the baseline model.

```python
# Calculate residuals for full model
y_pred = model.predict(X_scaled)
residuals_full = y - y_pred

# Baseline (mean-only) model
y_baseline = np.mean(y)
residuals_baseline = y - y_baseline

# Calculate Sum of Squares
SS_full = np.sum(residuals_full**2)  # Sum of Squares for Full Model
SS_baseline = np.sum(residuals_baseline**2)  # Sum of Squares for Baseline

# Degrees of freedom
df_full = len(y) - len(X.columns) - 1  # Full model df
df_baseline = len(y) - 1               # Baseline model df

# F-statistic
F_stat = ((SS_baseline - SS_full) / (len(X.columns))) / (SS_full / df_full)

# p-value
p_value_anova = 1 - t.cdf(F_stat, df_full)

# Display results
print("\nANOVA Results:")
print(f"F-statistic: {F_stat:.4f}")
print(f"p-value: {p_value_anova:.4f}")

if p_value_anova < 0.05:
    print("-> Reject the null hypothesis: The regression model is significant.")
else:
    print("-> Fail to reject the null hypothesis: The regression model is not significant.")
```
